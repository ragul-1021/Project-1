```python
import pandas as pd
import re
df = pd.read_csv("../data/spam.csv",
                 sep="\t",
                 header=None,
                 names=["label","message"]
                 )
def clean_text(text):
    text=text.lower()
    res=re.sub(r"[^a-zA-Z ]","",text)
    return res
df["message"] = df["message"].apply(clean_text)
df["message"]
```
